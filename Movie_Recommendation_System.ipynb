```python

import pandas as pd
import zipfile
import urllib.request
import os

# Download the MovieLens dataset
url = "https://files.grouplens.org/datasets/movielens/ml-latest-small.zip"
urllib.request.urlretrieve(url, "movies_dataset.zip")

# Extract the downloaded ZIP file
with zipfile.ZipFile("movies_dataset.zip", "r") as zip_ref:
    zip_ref.extractall(".")

# Load the movie information
movies = pd.read_csv("ml-latest-small/movies.csv")

# Display the first five movies
print("Dataset downloaded successfully!")
print("Total movies:", len(movies))
display(movies.head())
```

```python

# Check the dataset for missing values
print("Missing values:")
print(movies.isnull().sum())

# Remove any rows with missing information
movies = movies.dropna().reset_index(drop=True)

# Display the dataset information
print("\nDataset after cleaning:")
print("Total movies:", len(movies))

# Display some movie titles and genres
display(movies[['title', 'genres']].head(10))
```
